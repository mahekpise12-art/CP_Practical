## 4. Activity 2 – Build the Digit-Summation Loop

Now create an algorithm that repeatedly extracts the last digit and adds it to a running total.

For each iteration:

1. Extract the last digit.
2. Add it to `digit_sum`.
3. Remove the last digit from the number.
4. Continue until no digits remain.

### Example

For `5827`, your program should process:

```text

```

and finally produce:

```text

```

### Copilot Activity 2 – Loop Logic

Ask Copilot:

> Explain the logic of a while loop that repeatedly extracts the last digit of an integer, adds it to a running sum, and removes the last digit. Do not give me the complete Python code.

````python
# Activity 2
# Write a while-loop program to calculate the sum of digits.
number = 5827
sum_of_digits = 0
while number > 0:
    last_digit = number % 10
    sum_of_digits += last_digit
    number = number // 10
print("Sum_of_digits:",sum_of_digits)
````

## 6. Activity 4 – Convert a Chemical Batch Number into a Digit Sum

Now connect the basic algorithm with the syllabus application.

Interpret it simply as a batch identifier and calculate the sum of its digits.

### Important concept

If an identifier contains leading zeros, such as:

```text
001245
```

converting it directly to an integer will remove the leading zeros.

Therefore, when the value is an **identifier**, it is often better to store it as a string.

### Copilot Activity 3 – Number vs Identifier

Ask Copilot:

> Explain why a chemical batch number or sample ID should often be stored as a Python string instead of an integer, especially when it can contain leading zeros. Give a simple example. Do not write the complete program.

````python
# Activity 4
# Store a batch number as a string and calculate the sum of its digits.

batch_number = "5827"
sum_of_digits = 0
for digit in batch_number:
    sum_of_digits += int(digit)
print("Sum of digits in batch number:", sum_of_digits)
````

### Copilot Activity 4 – Processing Characters

Ask Copilot:

> I have a chemical batch identifier such as LAB001245. Explain how a Python for loop can inspect each character and identify which characters are digits. Mention the role of isdigit(). Do not write the complete program.

````python
# Activity 5
# Calculate the sum of only the digits contained in a batch identifier.
batch_number = "AB5827CD"
sum_of_digits = 0
for char in batch_number:
    if char.isdigit():
        sum_of_digits += int(char)
print("Sum of digits in batch identifier:", sum_of_digits)
````
