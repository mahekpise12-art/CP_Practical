## 2. Why Quadratic Equations Matter in Chemical Engineering

Quadratic equations are not being introduced here as an isolated mathematics exercise. They can arise when chemical-engineering relationships are rearranged to determine an unknown concentration, extent of reaction, or process variable.

In this practical we will use two contexts:

### Context A – Chemical Equilibrium
A simplified equilibrium relationship can lead to a quadratic equation in the **extent of reaction** `x`.

### Context B – Reaction Rate Expression
A rate expression containing a squared concentration can also be rearranged into a quadratic equation when the unknown concentration must be determined.

The objective is to teach students how to move through the complete chain:

**Chemical expression → algebraic equation → quadratic coefficients → Python calculation → roots → physical interpretation**

**Important:** The examples below are deliberately simplified for programming practice. They are not intended to replace a complete chemical-equilibrium or reaction-engineering analysis.

## 3. The Quadratic Formula

For an equation of the form:

```text
a x² + b x + c = 0
```

the roots are obtained from the quadratic formula.

The discriminant is:

```text
Δ = b² - 4ac
```

The discriminant tells us the nature of the mathematical roots:

| Discriminant | Roots |
|---|---|
| Δ > 0 | Two distinct real roots |
| Δ = 0 | One repeated real root |
| Δ < 0 | Two complex roots |

However, in chemical engineering **mathematical validity is not enough**. A root must also satisfy physical restrictions such as concentration ≥ 0 or extent of reaction lying within the allowable range.

## 5. Activity 2 – Chemical Equilibrium Problem

Consider the simplified equilibrium reaction:

```text
A + B ⇌ C
```

Assume initially:

- `[A]₀ = 1.0 mol/L`
- `[B]₀ = 1.0 mol/L`
- `[C]₀ = 0 mol/L`

Let `x` be the equilibrium concentration formed of C.

At equilibrium:

```text
[A] = 1 - x
[B] = 1 - x
[C] = x
```

For this simplified example, take:

```text
Kc = [C] / ([A][B])
```

Therefore:

```text
Kc = x / (1 - x)²
```

Rearranging gives:

```text
Kc x² - (2Kc + 1)x + Kc = 0
```

**This is the key chemical-engineering connection of this practical:** the equilibrium expression has been converted into a quadratic equation in `x`.

## 6. Activity 3 – Solve the Equilibrium Quadratic

Use:

```text
Kc = 4
```

For the derived equation:

```text
Kc x² - (2Kc + 1)x + Kc = 0
```

calculate the values of:

- `a`

### Engineering question

### Copilot Activity 3 – Physical Root Selection

Ask Copilot:

> In a chemical-equilibrium calculation, x represents an equilibrium concentration formed from an initial concentration of 1 mol/L. Explain what physical restrictions should be considered when choosing between mathematical roots. Do not solve the specific equation for me.

````python
# Activity 3
# For Kc = 4, calculate the quadratic coefficients and solve for x.
# Identify the physically meaningful root.

kc = 4
a = kc
b = -(2*kc + 1)
c = kc

determine = b**2 - 4*a*c
if determine < 0:
    print("No real roots.")
else:
    root1 = (-b + determine**0.5) / (2*a)
    root2 = (-b - determine**0.5) / (2*a)
    print("Roots:", root1, root2)
    
````
