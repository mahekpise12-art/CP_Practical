## 2. Chemical Engineering Context

In a chemical manufacturing plant, a quality-control laboratory checks a product batch before it is released. Measurements such as **purity, pH and moisture content** are compared with predefined quality specifications.

Based on these measurements, the batch may be classified as **Grade A, Grade B, Grade C**, or **Rejected**.

In this practical, you will build a simple Python-based quality-check system for such a situation.


## 3. Quality Specifications

For this educational exercise, use the following specifications:

| Parameter | Grade A | Grade B | Grade C | Reject |
|---|---|---|---|---|
| Purity | ≥ 98% | 95–97.99% | 90–94.99% | < 90% |
| pH | 6.5–7.5 | 6.0–8.0 | 5.0–9.0 | Outside 5–9 |
| Moisture | ≤ 1% | ≤ 2% | ≤ 4% | > 4% |

**Important:** These values are created only for this practical. They are not specifications for a real chemical product.

## 5. Activity 2 – Add pH Checking

Now extend your program so that pH is also considered.

For a batch to receive **Grade A**, both conditions must be satisfied:

- Purity ≥ 98%
- pH between 6.5 and 7.5

If the purity is excellent but the pH is outside the Grade A range, the product must not be classified as Grade A.

### Think before coding

What logical operator is required when **both conditions must be true**?


### Copilot Activity 2 – Understand `and`

Ask Copilot:

> Explain the Python `and` operator using the example of checking whether the pH of a chemical batch is between 6.5 and 7.5. Give examples, but let me write the final code myself.

Then modify your program to include pH.


````python
# Activity 2
# Add pH as a quality parameter.
# Write/modify your code here.
purity = float(input("Enter the purity of the product (as a percentage): ")) 
print(f"Purity entered: {purity}%")
pH = float(input("Enter the pH of the product: "))
print(f"pH entered: {pH}")
if purity >= 98.0 and 6.5 <= pH <= 7.5:
    print("Product is Grade A.")
elif purity >= 95.0 and 6.0 <= pH <= 8.0:
    print("Product is Grade B.")
elif purity >= 90.0 and 5.0 <= pH <= 9.0:
    print("Product is Grade C.")
else:
    print("Product is Rejected.")
````

## 6. Activity 3 – Add Moisture Content

Now include moisture content.

For Grade A, all three conditions must be satisfied:

1. Purity ≥ 98%
2. pH between 6.5 and 7.5
3. Moisture ≤ 1%

For this activity, create your own logical structure for determining the final grade.

**Do not immediately ask Copilot for the complete answer. First draw or write the decision logic yourself.**


````python
# Activity 3
# Complete chemical quality grading using:
# purity, pH and moisture.
# Write your solution here.
purity = float(input("Enter the purity of the product (as a percentage): ")) 
print(f"Purity entered: {purity}%")
pH = float(input("Enter the pH of the product: "))
print(f"pH entered: {pH}")
moisture = float(input("Enter the moisture content of the product (as a percentage): "))
print(f"Moisture entered: {moisture}%")
if purity >= 98.0 and 6.5 <= pH <= 7.5 and moisture <= 1.0:
    print("Product is Grade A.")
elif purity >= 95.0 and 6.0 <= pH <= 8.0 and moisture <= 2.0:
    print("Product is Grade B.")
elif purity >= 90.0 and 5.0 <= pH <= 9.0 and moisture <= 4.0:
    print("Product is Grade C.")
else:
    print("Product is Rejected.")
````

## 7. Activity 4 – Input Validation

A real program should not accept obviously invalid measurements.

For this exercise, use these basic checks:

- Purity must be between 0 and 100.
- pH must be between 0 and 14.
- Moisture must not be negative.

If an invalid value is entered, display an appropriate message instead of performing the quality grading.

### Copilot challenge

Ask Copilot:

> I need to validate purity, pH and moisture inputs in a beginner-level Python program. Explain a simple approach using if statements. Do not rewrite my entire program.


````python
# Activity 4
# Add input validation to your quality-check program.
purity = float(input("Enter the purity of the product (as a percentage): ")) 
print(f"Purity entered: {purity}%")
if purity < 0 or purity > 100:
    print("Invalid purity value. Please enter a value between 0 and 100.")
    exit(1)
pH = float(input("Enter the pH of the product: "))
print(f"pH entered: {pH}")
if pH < 0 or pH > 14:
    print("Invalid pH value. Please enter a value between 0 and 14.")
    exit(1)
moisture = float(input("Enter the moisture content of the product (as a percentage): "))
print(f"Moisture entered: {moisture}%")
if moisture < 0 or moisture > 100:
    print("Invalid moisture value. Please enter a value between 0 and 100.")
    exit(1)
if purity >= 98.0 and 6.5 <= pH <= 7.5 and moisture <= 1.0:
    print("Product is Grade A.")
elif purity >= 95.0 and 6.0 <= pH <= 8.0 and moisture <= 2.0:
    print("Product is Grade B.")
elif purity >= 90.0 and 5.0 <= pH <= 9.0 and moisture <= 4.0:
    print("Product is Grade C.")
else:
    print("Product is Rejected.")
````
